## 3. Docstrings y Documentación

```python
# Docstring de una línea
def sumar(a, b):
    """Retorna la suma de dos números."""
    return a + b

# Docstring multilínea (estilo Google)
def calcular_estadisticas(datos, incluir_mediana=True):
    """
    Calcula estadísticas descriptivas de una lista de números.
    
    Args:
        datos: Lista de números para analizar.
        incluir_mediana: Si True, incluye la mediana en el resultado.
            Por defecto es True.
    
    Returns:
        Diccionario con las estadísticas calculadas:
        - 'media': Promedio aritmético
        - 'minimo': Valor mínimo
        - 'maximo': Valor máximo
        - 'mediana': Valor central (si incluir_mediana=True)
    
    Raises:
        ValueError: Si la lista está vacía.
        TypeError: Si los elementos no son numéricos.
    
    Example:
        >>> calcular_estadisticas([1, 2, 3, 4, 5])
        {'media': 3.0, 'minimo': 1, 'maximo': 5, 'mediana': 3}
    """
    if not datos:
        raise ValueError("La lista no puede estar vacía")
    
    resultado = {
        'media': sum(datos) / len(datos),
        'minimo': min(datos),
        'maximo': max(datos)
    }
    
    if incluir_mediana:
        ordenados = sorted(datos)
        n = len(ordenados)
        mid = n // 2
        resultado['mediana'] = ordenados[mid] if n % 2 else (ordenados[mid-1] + ordenados[mid]) / 2
    
    return resultado

# Ver la documentación
help(calcular_estadisticas)
```

## 4. Type Hints y Anotaciones

```python
# Type hints básicos
from typing import List, Dict, Optional, Union, Tuple, Callable

# Variables con tipo
nombre: str = "Juan"
edad: int = 30
precio: float = 19.99
activo: bool = True

# Funciones con tipos
def saludar(nombre: str) -> str:
    return f"Hola, {nombre}"

def sumar_lista(numeros: List[int]) -> int:
    return sum(numeros)

def obtener_usuario(id: int) -> Dict[str, str]:
    return {"id": str(id), "nombre": "Usuario"}

print(saludar("Ana"))
print(sumar_lista([1, 2, 3, 4, 5]))
```

```python
# Tipos opcionales y uniones
from typing import Optional, Union

# Optional[X] es equivalente a Union[X, None]
def buscar_usuario(id: int) -> Optional[Dict[str, str]]:
    """Retorna el usuario o None si no existe."""
    usuarios = {1: {"nombre": "Ana"}, 2: {"nombre": "Bob"}}
    return usuarios.get(id)

# Union para múltiples tipos
def procesar_entrada(valor: Union[str, int, float]) -> str:
    """Acepta string, int o float."""
    return str(valor)

# Python 3.10+: Sintaxis simplificada
def procesar_moderno(valor: str | int | float) -> str:
    return str(valor)

print(buscar_usuario(1))
print(buscar_usuario(99))
```

```python
# Tipos genéricos y avanzados
from typing import TypeVar, Generic, Callable, Any

# TypeVar para genéricos
T = TypeVar('T')

def primer_elemento(lista: List[T]) -> T:
    """Retorna el primer elemento de cualquier lista."""
    return lista[0]

# Callable para funciones como parámetro
def aplicar_funcion(
    datos: List[int], 
    transformar: Callable[[int], int]
) -> List[int]:
    """Aplica una función a cada elemento."""
    return [transformar(x) for x in datos]

# Uso
numeros = [1, 2, 3, 4, 5]
cuadrados = aplicar_funcion(numeros, lambda x: x**2)
print(f"Cuadrados: {cuadrados}")
```

```python
# Ejercicio 2: Escribir docstrings completos
from typing import List, Optional

def filtrar_y_transformar(
    datos: List[int],
    minimo: int = 0,
    maximo: Optional[int] = None,
    transformacion: str = "identidad"
) -> List[int]:
    """
    Filtra una lista de enteros y aplica una transformación.
    
    Filtra los elementos que están dentro del rango [minimo, maximo]
    y aplica la transformación especificada a cada elemento.
    
    Args:
        datos: Lista de enteros a procesar.
        minimo: Valor mínimo permitido (inclusive). Por defecto 0.
        maximo: Valor máximo permitido (inclusive). Si es None,
            no hay límite superior.
        transformacion: Tipo de transformación a aplicar.
            Opciones: "identidad", "cuadrado", "doble".
    
    Returns:
        Lista de enteros filtrados y transformados.
    
    Raises:
        ValueError: Si la transformación no es válida.
    
    Example:
        >>> filtrar_y_transformar([1, 2, 3, 4, 5], minimo=2, maximo=4)
        [2, 3, 4]
        >>> filtrar_y_transformar([1, 2, 3], transformacion="cuadrado")
        [1, 4, 9]
    """
    transformaciones = {
        "identidad": lambda x: x,
        "cuadrado": lambda x: x ** 2,
        "doble": lambda x: x * 2
    }
    
    if transformacion not in transformaciones:
        raise ValueError(f"Transformación inválida: {transformacion}")
    
    func = transformaciones[transformacion]
    resultado = []
    
    for valor in datos:
        if valor >= minimo and (maximo is None or valor <= maximo):
            resultado.append(func(valor))
    
    return resultado

# Pruebas
print(filtrar_y_transformar([1, 2, 3, 4, 5], minimo=2, maximo=4))
print(filtrar_y_transformar([1, 2, 3], transformacion="cuadrado"))
```

```python
# Ejercicio 3: Clase con type hints completos
from typing import List, Dict, Optional
from dataclasses import dataclass, field
from datetime import datetime

@dataclass
class Tarea:
    """
    Representa una tarea en un sistema de gestión.
    
    Attributes:
        id: Identificador único de la tarea.
        titulo: Título descriptivo.
        descripcion: Descripción detallada.
        completada: Estado de completitud.
        fecha_creacion: Momento de creación.
        etiquetas: Lista de etiquetas asociadas.
    """
    id: int
    titulo: str
    descripcion: str = ""
    completada: bool = False
    fecha_creacion: datetime = field(default_factory=datetime.now)
    etiquetas: List[str] = field(default_factory=list)
    
    def completar(self) -> None:
        """Marca la tarea como completada."""
        self.completada = True
    
    def agregar_etiqueta(self, etiqueta: str) -> None:
        """Agrega una etiqueta a la tarea."""
        if etiqueta not in self.etiquetas:
            self.etiquetas.append(etiqueta)


class GestorTareas:
    """Gestiona una colección de tareas."""
    
    def __init__(self) -> None:
        self._tareas: Dict[int, Tarea] = {}
        self._siguiente_id: int = 1
    
    def crear_tarea(self, titulo: str, descripcion: str = "") -> Tarea:
        """Crea y registra una nueva tarea."""
        tarea = Tarea(id=self._siguiente_id, titulo=titulo, descripcion=descripcion)
        self._tareas[self._siguiente_id] = tarea
        self._siguiente_id += 1
        return tarea
    
    def obtener_tarea(self, id: int) -> Optional[Tarea]:
        """Obtiene una tarea por su ID."""
        return self._tareas.get(id)
    
    def listar_pendientes(self) -> List[Tarea]:
        """Lista todas las tareas no completadas."""
        return [t for t in self._tareas.values() if not t.completada]

# Uso
gestor = GestorTareas()
tarea1 = gestor.crear_tarea("Aprender PEP8", "Estudiar convenciones")
tarea1.agregar_etiqueta("python")
print(tarea1)
```

```python
# Ejercicio 2: Escribir docstring completo para esta función

def procesar_datos(datos, filtro=None, ordenar=True, limite=None):
    resultado = datos.copy()
    if filtro:
        resultado = [d for d in resultado if filtro(d)]
    if ordenar:
        resultado.sort()
    if limite:
        resultado = resultado[:limite]
    return resultado

# Reescribe la función con docstring completo estilo Google
# ...
def process_data(data, filters=None, sort=True, limit=None):
    """
    Processes data and filters, sorts and limits the results.

    Args:
        data (list): List of elements to process.
        filters (callable, optional): if there is a filter applies it to the data processing task.
            None by default.
        sort (bool, optional): if True, it sorts the data following sort requirements, if False the returned data will be unsorted
            True by default.
        limit (int, optional): if differente than None, should limit the amount of results returned.
            None by default.

    Returns:
        List with the result of the data processing
    """
    result = data.copy()
    if filters:
        result = [d for d in result if filters(d)]
    if sort:
        result.sort()
    if limit:
        result = result[:limit]
    return result
```
